```python
import numpy as np
import pandas as pd
df = pd.read_csv("data/raw/vehicles.csv")

print(df.head())

print(df.info())

print(df.describe())

columns = [
    "price",
    "year",
    "manufacturer",
    "model",
    "fuel",
    "transmission",
    "odometer",
    "condition"
]

df = df[columns]

df = df.dropna()

df = df[df["price"] > 0]

df = df[df["odometer"] > 0]

df = df.drop_duplicates()

df.to_csv("data/cleaned/cars_cleaned.csv", index=False)

print("Data saved done!")
```
